#Stop Words :Stop words are common words that are often removed from text because they carry little useful information for certain NLP tasks.

- Examples:
"The cat is sitting on the table."

- After removing them: "cat sitting table"

```
| Category      | Examples               |
| ------------- | ---------------------- |
| Articles      | a, an, the             |
| Pronouns      | I, you, he, she, they  |
| Prepositions  | in, on, at, by, for    |
| Conjunctions  | and, or, but           |
| Helping verbs | is, am, are, was, were |
| Others        | this, that, very, with |
```

#**Stemming** :Stemming is the process of reducing a word to its root/base form by removing prefixes or suffixes.
3 types :
 1. Porter Stemming :
 2. Snowball Stemming :
 3. Lancaster Stemming :

#Stemming vs Lemmatization : This is important for NLP interviews:

Stemming → faster, rule-based, may produce meaningless words.

Lemmatization → slower, uses linguistic knowledge, usually produces a valid dictionary word.

Example:
- studies → studi      # stemming
- studies → study      # lemmatization

#1. Porter Stemmer
- One of the most commonly used stemmers.
- Uses a sequence of rules to remove suffixes.
- Balanced between speed and accuracy.
- Example:
  - playing → play
  - studies → studi
```
from nltk.stem import PorterStemmer

ps = PorterStemmer()
print(ps.stem("playing"))
```


#2. Snowball Stemmer
- An improved and more flexible version of the Porter algorithm.
- Also called Porter2 for English.
- Supports multiple languages.
- Generally considered more consistent than Porter.
```
from nltk.stem import SnowballStemmer

ss = SnowballStemmer("english")
print(ss.stem("playing"))
```

#3. Lancaster Stemmer
- Uses a more aggressive stemming approach.
- Can reduce words much more heavily.
- Faster, but sometimes produces stems that are difficult to interpret.
```
from nltk.stem import LancasterStemmer

ls = LancasterStemmer()
print(ls.stem("playing"))
```


```
Word          Snowball       Lancaster
-----------------------------------------
maximum       maximum        maxim
organization  organ         org
happily       happili       happy
friendliness  friend        friend
```

#Sentiment analysis model

````python
import numpy as np
import pandas as pd
import re
import nltk
from sklearn.naive_bayes import MultinomialNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
from sklearn.feature_extraction.text import TfidfVectorizer
````

````python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("lakshmi25npathi/imdb-dataset-of-50k-movie-reviews")

print("Path to dataset files:", path)
````

````python
df=pd.read_csv(f'{path}/IMDB Dataset.csv')
df.head()
````

````python
df.shape
````

````python
df['review'][0]
````

````python
def clean(text):
     text = re.sub(r'<.*?>','',text)  #remove the html tags
     text = re.sub(r'[^a-zA-Z\s]','',text) # reemove the spectial character and number
     text = re.sub(r'\s+',' ',text) #remove the unwanted space before and after the word
     text = text.lower() # convert into small letter
     return text
````

````python
df['review']=df['review'].apply(clean)
````

````python
df['review'][0]
````

````python
y=df['sentiment']
x=df['review']
````

````python
vector= TfidfVectorizer(max_features=5000)
x_vec=vector.fit_transform(x)
````

````python
x_train, x_test, y_train, y_test= train_test_split(x_vec, y, test_size=0.2, random_state=42)
````

````python
model=MultinomialNB()
model.fit(x_train, y_train)
````

````python
predict=model.predict(x_test)
predict
````

````python
scr=classification_report(y_test, predict)
print(scr)
````

````python
import pickle
pickle.dump(model,open('Sentimonster.pkl', 'wb'))
pickle.dump(vector, open('vectormonster.pkl', 'wb'))
print("Both model created successfully")
````
